```python
import pandas as pd

data = pd.read_csv(r"PlayStation Sales and Metadata (PS3PS4PS5) (Oct 2025).csv")

data.head()
```

```python
data.info()
```

```python
# Standardizing the column names

data.columns = data.columns.str.strip().str.replace(' ', '_').str.lower() 


# Removing column duplcates

data.drop_duplicates(inplace=True)


# Removing Null values

# Replacing NaN values with Unkown

columns = ['game', 'console', 'name', 'publisher', 'developer', 'genres', 'platforms']

for col in columns:
	if col in data.columns:
		data[col] = data[col].fillna('unknown')

# Replacing NaN values with numerical values

columns = ['total_shipped', 'total_sales', 'na_sales', 'pal_sales', 'japan_sales', 'other_sales', 'rating', 'ratings_count', 'metacritic']


for col in columns:
	if col in data.columns:
		data[col] = pd.to_numeric(data[col], errors='coerce') # Forcing the values to be Numeric
		# selecting columns that values missing should be -1
		if col in ['rating', 'ratings_count', 'metacritic']:
			data[col] = data[col].fillna(-1)  # -1 = missing. 0 mean score 0 not missing
		else:
			data[col] = data[col].fillna(0)


# Replacing Null values for Date Columns

columns = ['release_date', 'last_update']

for col in columns:
	if col in data.columns:
		data[col] = pd.to_datetime(data[col], errors = 'coerce')
		data[col] = data[col].fillna(pd.NaT)
```
